```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("MIMIC-IV Readmission").getOrCreate()

# Fast reload from parquet
admissions = spark.read.parquet("gs://ba865-t9-mimiciv/parquet/admissions")
patients = spark.read.parquet("gs://ba865-t9-mimiciv/parquet/patients")
labevents = spark.read.parquet("gs://ba865-t9-mimiciv/parquet/labevents")
diagnoses = spark.read.parquet("gs://ba865-t9-mimiciv/parquet/diagnoses")
icustays = spark.read.parquet("gs://ba865-t9-mimiciv/parquet/icustays")
chartevents = spark.read.parquet("gs://ba865-t9-mimiciv/parquet/chartevents")
cohort = spark.read.parquet("gs://ba865-t9-mimiciv/parquet/cohort")

print("Loaded!")
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# Sort admissions per patient by admit time
window = Window.partitionBy("subject_id").orderBy("admittime")

admissions_labeled = admissions.withColumn(
    "next_admittime",
    F.lead("admittime").over(window)
).withColumn(
    "admittime", F.to_timestamp("admittime")
).withColumn(
    "dischtime", F.to_timestamp("dischtime")
).withColumn(
    "next_admittime", F.to_timestamp("next_admittime")
).withColumn(
    "days_to_readmit",
    F.datediff(F.col("next_admittime"), F.col("dischtime"))
).withColumn(
    "readmitted_30d",
    F.when(F.col("days_to_readmit") <= 30, 1).otherwise(0)
)

# Exclude in-hospital deaths and transfers
cohort = admissions_labeled.filter(
    (F.col("hospital_expire_flag") == 0) &
    (~F.col("discharge_location").isin("DIED", "HOSPICE"))
)

cohort.groupBy("readmitted_30d").count().show()
```

```python
# join patient demographics
cohort = cohort.join(
    patients.select("subject_id", "anchor_age", "gender"),
    on="subject_id",
    how="left"
)
cohort.select("subject_id", "hadm_id", "anchor_age", "gender", "readmitted_30d").show(5)
```

```python
# calculate length of stay in days
cohort = cohort.withColumn(
    "los_days",
    F.round(
        F.unix_timestamp("dischtime") / 86400 - F.unix_timestamp("admittime") / 86400,
        2
    )
)

# prior admissions count (admissions before this one, per patient)
prior_window = Window.partitionBy("subject_id").orderBy("admittime")
cohort = cohort.withColumn(
    "prior_admissions",
    (F.row_number().over(prior_window) - 1)
)

cohort.select("subject_id", "hadm_id", "los_days", "prior_admissions").show(5)
```

```python
# keep only the primary diagnosis (seq_num == 1) and extract ICD-10 chapter (first char)
primary_dx = diagnoses.filter(F.col("seq_num") == 1) \
    .select(
        "hadm_id",
        F.substring(F.col("icd_code"), 1, 1).alias("icd10_chapter")
    )

cohort = cohort.join(primary_dx, on="hadm_id", how="left")
cohort.select("hadm_id", "icd10_chapter").show(5)
```

```python
# LOINC itemids in MIMIC-IV:
# 50912 = Creatinine, 51222 = Hemoglobin
LAB_ITEMS = {50912: "creatinine", 51222: "hemoglobin"}

labs_filtered = labevents.filter(F.col("itemid").isin(list(LAB_ITEMS.keys()))) \
    .select("hadm_id", "itemid", "valuenum", "charttime")

# Take the last recorded value per hadm_id per itemid
lab_window = Window.partitionBy("hadm_id", "itemid").orderBy(F.desc("charttime"))

labs_last = labs_filtered \
    .withColumn("rn", F.row_number().over(lab_window)) \
    .filter(F.col("rn") == 1) \
    .withColumn("lab_name", F.when(F.col("itemid") == 50912, "creatinine").otherwise("hemoglobin")) \
    .groupBy("hadm_id").pivot("lab_name").agg(F.first("valuenum"))

cohort = cohort.join(labs_last, on="hadm_id", how="left")
cohort.select("hadm_id", "creatinine", "hemoglobin").show(5)
```

```python
# itemid 220045 = Heart Rate in MIMIC-IV chartevents
hr = chartevents.filter(F.col("itemid") == 220045) \
    .select("hadm_id", "valuenum", "charttime")

hr_window = Window.partitionBy("hadm_id").orderBy(F.desc("charttime"))

hr_last = hr \
    .withColumn("rn", F.row_number().over(hr_window)) \
    .filter(F.col("rn") == 1) \
    .select("hadm_id", F.col("valuenum").alias("heart_rate_discharge"))

cohort = cohort.join(hr_last, on="hadm_id", how="left")
cohort.select("hadm_id", "heart_rate_discharge").show(5)
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window

spark = SparkSession.builder.appName("MIMIC-IV Readmission").getOrCreate()

cohort_structured = spark.read.parquet("gs://ba865-t9-mimiciv/parquet/cohort_structured")
discharge_notes   = spark.read.parquet("gs://ba865-t9-mimiciv/parquet/discharge_notes")

print("Cohort rows   :", cohort_structured.count())
print("Notes rows    :", discharge_notes.count())
cohort_structured.printSchema()
```

```python
# One discharge summary per admission — take the latest if multiple exist
note_window = Window.partitionBy("hadm_id").orderBy(F.desc("charttime"))

ds_notes = discharge_notes \
    .filter(F.col("note_type") == "DS") \
    .withColumn("rn", F.row_number().over(note_window)) \
    .filter(F.col("rn") == 1) \
    .select("hadm_id", F.col("text").alias("discharge_text"))

cohort_final = cohort_structured.join(ds_notes, on="hadm_id", how="inner")

print(f"Before join : {cohort_structured.count()}")
print(f"After join  : {cohort_final.count()}")
cohort_final.groupBy("readmitted_30d").count().show()

# Save so we never have to rejoin again
cohort_final.write.mode("overwrite").parquet("gs://ba865-t9-mimiciv/parquet/cohort_final")
```

```python
# Select only the columns we need for modeling
model_cols = [
    "hadm_id", "readmitted_30d",
    "anchor_age", "gender", "los_days", "prior_admissions",
    "icd10_chapter", "admission_type", "insurance",
    "creatinine", "hemoglobin", "heart_rate_discharge",
    "discharge_text"
]

cohort_final.select(model_cols).write.mode("overwrite").parquet("gs://ba865-t9-mimiciv/parquet/cohort_model_input")

print("Saved! Now reading directly with pandas...")
import pandas as pd
df = pd.read_parquet("gs://ba865-t9-mimiciv/parquet/cohort_model_input")

print("Shape:", df.shape)
print(df["readmitted_30d"].value_counts())
df.head(3)
```

```python
import subprocess
subprocess.run(["pip", "install", "-q", "keras_hub", "keras"], check=True)

import keras_hub
import keras
from keras import layers
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split

print("Keras version    :", keras.__version__)
print("keras_hub version:", keras_hub.__version__)
```

```python
# Encode categoricals
for col in ["gender", "icd10_chapter", "admission_type", "insurance"]:
    df[col] = LabelEncoder().fit_transform(df[col].astype(str))

structured_cols = [
    "anchor_age", "gender", "los_days", "prior_admissions",
    "icd10_chapter", "creatinine", "hemoglobin", "heart_rate_discharge",
    "admission_type", "insurance"
]

scaler   = StandardScaler()
X_struct = scaler.fit_transform(df[structured_cols].fillna(0))
y        = df["readmitted_30d"].values
texts    = df["discharge_text"].fillna("").tolist()

# Split at patient level to avoid data leakage
idx = np.arange(len(df))
train_idx, test_idx = train_test_split(idx, test_size=0.2, random_state=42, stratify=y)

X_struct_train, X_struct_test = X_struct[train_idx], X_struct[test_idx]
texts_train = [texts[i] for i in train_idx]
texts_test  = [texts[i] for i in test_idx]
y_train, y_test = y[train_idx], y[test_idx]

print(f"Train: {len(y_train)} | Test: {len(y_test)}")
print(f"Readmission rate (train): {y_train.mean():.3f}")
```

```python
tokenizer = keras_hub.models.Tokenizer.from_preset("roberta_base_en")
backbone  = keras_hub.models.Backbone.from_preset("roberta_base_en")

# Freeze backbone — use as feature extractor, not fine-tuning (same as class 8.4)
backbone.trainable = False

backbone.summary()
```

```python
import tensorflow as tf

def preprocess_text(text, label):
    # Same StartEndPacker pattern as class 8.4
    packer = keras_hub.layers.StartEndPacker(
        sequence_length=512,
        start_value=tokenizer.start_token_id,
        end_value=tokenizer.end_token_id,
        pad_value=tokenizer.pad_token_id,
        return_padding_mask=True,
    )
    token_ids, padding_mask = packer(tokenizer(text))
    return {"token_ids": token_ids, "padding_mask": padding_mask}, label

BATCH_SIZE = 16

train_text_ds = tf.data.Dataset.from_tensor_slices((texts_train, y_train)) \
    .map(preprocess_text).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

test_text_ds = tf.data.Dataset.from_tensor_slices((texts_test, y_test)) \
    .map(preprocess_text).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

# Preview — same check as class 8.4
print(next(iter(train_text_ds)))
```

```python
# ── Text branch (class 8.4 pattern) ──────────────────────────────
text_inputs = backbone.input          # token_ids + padding_mask
x_text = backbone(text_inputs)
x_text = x_text[:, 0, :]             # CLS token = document-level embedding
x_text = layers.Dropout(0.1)(x_text)
x_text = layers.Dense(256, activation="relu")(x_text)
x_text = layers.Dropout(0.1)(x_text)

# ── Structured branch (class 6.2 pattern) ────────────────────────
struct_input = layers.Input(shape=(X_struct.shape[1],), name="structured")
x_struct = layers.Dense(64, activation="relu")(struct_input)
x_struct = layers.Dense(32, activation="relu")(x_struct)

# ── Merge with Concatenate (exactly as class 6.2) ─────────────────
merged = layers.Concatenate()([x_text, x_struct])
x = layers.Dense(64, activation="relu")(merged)
x = layers.Dropout(0.2)(x)
output = layers.Dense(1, activation="sigmoid")(x)

model = keras.Model(
    inputs=text_inputs + [struct_input],
    outputs=output
)

model.compile(
    optimizer=keras.optimizers.Adam(5e-5),
    loss="binary_crossentropy",
    metrics=["AUC", "accuracy"]
)

keras.utils.plot_model(model, show_shapes=True, dpi=60)
```

```python
# Zip text + structured datasets together
train_struct_ds = tf.data.Dataset.from_tensor_slices(
    (X_struct_train.astype("float32"), y_train)
).batch(BATCH_SIZE)

def combine(text_batch, struct_batch):
    (text_inputs, labels), (struct_inputs, _) = text_batch, struct_batch
    return {**text_inputs, "structured": struct_inputs}, labels

train_ds = tf.data.Dataset.zip((train_text_ds, train_struct_ds)).map(combine)

history = model.fit(
    train_ds,
    epochs=5,
    class_weight={0: 1, 1: 3}  # readmissions are ~15-20% — handle imbalance
)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score, confusion_matrix, ConfusionMatrixDisplay

# Plot AUC curve
plt.plot(history.history["AUC"], label="Train AUC")
plt.xlabel("Epoch"); plt.ylabel("AUC"); plt.legend(); plt.title("Training AUC")
plt.show()

# Build test combined dataset
test_struct_ds = tf.data.Dataset.from_tensor_slices(
    (X_struct_test.astype("float32"), y_test)
).batch(BATCH_SIZE)
test_ds = tf.data.Dataset.zip((test_text_ds, test_struct_ds)).map(combine)

# Predict & evaluate
y_pred     = model.predict(test_ds).flatten()
y_pred_bin = (y_pred > 0.5).astype(int)

print(f"Test AUC: {roc_auc_score(y_test, y_pred):.4f}")

cm = confusion_matrix(y_test, y_pred_bin)
ConfusionMatrixDisplay(cm, display_labels=["Not Readmitted", "Readmitted"]).plot()
plt.show()
```
